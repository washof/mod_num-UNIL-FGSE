# Tutorial 1: Se familiariser avec l'approximation d'un dérivée spatiale

Pour résoudre un modèle (p.e. de diffusion), il nous faudra trés souvent dans ces modèles appoximer des dérivées, c.a.d. de facçon numérique (pas de façon exacte/analytique). Pour illustrer cela, nous définisons un vecteur $F$ avec la fonction sinus, sa dérivée "exacte" définit avec la fonction cosinus, et sa dérivée "approchée" défini au moyen d'une différence finie (approximation que nous utiliserons à mainte reprise dans ce cours). Le code suivant fait tout cela, et montre que la dérivée approximative est trés proche de la dérivée exacte:

````python
import numpy as np
import matplotlib.pyplot as plt

# Define x and dx
nx = 100
x = np.linspace(0, 2 * np.pi, nx)  # Example range of x
dx = x[1] - x[0]  # Compute dx

# Compute F, dFdx_exact, and dFdx_appro
F = np.sin(x)
dFdx_exact = np.cos(x)
dFdx_appro = (F[1:] - F[:-1]) / dx

# Compute midpoints (xm)
xm = (x[1:] + x[:-1]) / 2

# Plotting
plt.plot(x, dFdx_exact, label="dF/dx exact (blue)", color="blue")
plt.plot(xm, dFdx_appro, label="dF/dx approx (red)", color="red", linestyle="--")

# Add labels, legend, and title
plt.xlabel("x")
plt.ylabel("dF/dx")
plt.title("Comparaison entre dérivées Exact et Approximative")
plt.legend()

# Display the plot
plt.grid(True)
plt.show()
````

Notons que pour afficher les deux `dFdx_exact` et `dFdx_approx` sur un meme graphique, il a fallue calculer les points au centre des cellules `xm = (x[1:] + x[:-1]) / 2` cat la dérivée approximative est de taille `nx-1` puisque l'on perd une cellule lors de la différence finie `(F[1:] - F[:-1]) / dx`, il faut donc "plotter" `dFdx_approx` par rapport à `xm` et non `x` qui est de taille `nx`.

````python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output, display
 
# Physical parameters
D = 0.8     # diffusivite
Lx = 30    # longueur of model
C_l = 500   # Condition de bord gauche
C_r = 0   # Condition de bord droit
ttot = 20  # temps total

# Numerical parameters
nx = 200                      # nombre de cellules
dx = Lx / (nx - 1)            # taille d'une cellule
x = np.linspace(0, Lx, nx)    # coordonees des point du maillage
dt = dx**2 / (2.1 * D)        # condition sur le pas de temps
nt = int(ttot/dt)                 # nombre de pas de temps
nout = 100                    # afficher la figure chaque 100e step

# Initial conditions
# CORRECTEUR: Your initialization is fine, but it's clearer and more general to define the array with C_r before overwriting part of it.
# C = np.zeros(nx)              # CORRECTEUR: not wrong, but less general
C = np.ones(nx) * C_r
C[:int(nx/2)] = 500        # modification de la conc sur la gauche du plot

# create the figure outside the loop 
fig, ax = plt.subplots()

# loop
for it in range(nt):
    # Mettre a jour le flux q en fonction de C 
    q = - D * ( C[1:] - C[:-1] ) / dx
    # Mettre a jour dCdt
    dCdt = - ( q[1:] - q[:-1] ) / dx
    # Mettre a jour C
    C[1:-1] += dCdt * dt

    # plot
    if it % nout == 0:
        clear_output(wait=True) 
        ax.cla() 
        ax.plot(x, C)
        ax.set_xlabel('Distance')
        ax.set_ylabel('Concentration')
        display(fig)
        plt.pause(0.1)
````

- Vous effectuerez un changement des conditions aux frontières. Jusqu'à présent, les conditions aux frontières imposaient implicitement une concentration inchangée au bord, c'est-à-dire les mêmes que les conditions initiales étant donné que les valeurs de la solution aux bords ne sont pas mises à jour. Ici, la condition de bord $C = 0$ est une condition dite de "Dirichlet". Une autre condition de bord (dite de “Neumann”) impose $\frac{\partial C}{\partial x} = 0$, c'est-à-dire qu'il n'y a plus d'échange vers l'extérieur, en forçant un flux nul aux bords (en raison de l'équation (2)). Numériquement parlant, cela revient à imposer que les deux cellules de bord aient la même valeur. Dans le cas présent, on suppose une molasse imperméable qui rend tout flux vers l'extérieur impossible à l'extrémité gauche, et que la concentration à l'extrémité droite du modèle doit être maintenue à zéro afin que tout le polluant puisse quitter le domaine. On implémente ces deux conditions ainsi en Python dans la boucle temporelle après la mise à jour de la diffusion :

```python

Cependant, la commande `q_old = q` ne sera pas efficace en Python, car `q_old = q` changera automatiquement lorsque `q` sera mis à jour (une spécificité de Python). En Python, il faut créer une "copie" explicitement avec la commande `q_old = np.copy(q)`, qui permettra de créer une nouvelle variable `q_old` retenant l'ancienne valeur de `q`, même après que celle-ci soit mise à jour.

````python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output, display
 
# Physical parameters
D = 0.001     # diffusivite
Lx = 300    # longueur of model
C_ini = 2000   # Condition de bord gauche
ttot = 200 * 86400  # temps total (= 200 jours * 86400 secondes par jour)

# Numerical parameters
nx = 300                      # nombre de cellules
dx = Lx / (nx - 1)            # taille d'une cellule
x = np.linspace(0, Lx, nx)    # coordonees des point du maillage
dt = dx**2 / (2.1 * D)        # condition sur le pas de temps
nt = int(ttot/dt)             # nombre de pas de temps
nout = 1000                   # afficher la figure chaque 500e step
venoge = 250                  # position de la venoge dans le modèle
x_init  = [30, 50]            # CORRECTEUR: store the position of the train to avoid hard coding later
C_thr = 0.1                   # CORRECTEUR: Add an explicit concentration threshold variable (used below instead of hard-coding 0.1).
d2s = 60 * 60 * 24            # CORRECTEUR: Add seconds-per-day to report times in days 

# Initial conditions
C = np.zeros(nx)           # conc sur toute la largueur du plot
#C[30:50] = C_ini          # CORRECTEUR: you need to convert the spatial positions to indices
C[int(x_init[0]/dx):int(x_init[1]/dx)] = C_ini # CORRECTEUR: use the stored train position variable (x_init) and convert to indices

q = np.zeros(nx-1)         # inititalisation de q pour avoir une valeur cohérente dans la boucle if du flux
time = 0 # CORRECTEUR: initialize time

# Flags pour bloquer les boucles if
flag_conc = True
flag_flux = True

# create the figure outside the loop 
fig, ax = plt.subplots()

# loop
for it in range(nt):
    time += dt  # CORRECTEUR: track time
    q_old = np.copy(q) # copie de q de la dernière itération pour calcul de temps baisse flux

    # Equations de flux
    q = - D * ( C[1:] - C[:-1] ) / dx # Mettre a jour le flux q en fonction de C 
    dCdt = - ( q[1:] - q[:-1] ) / dx # Mettre a jour dCdt
    C[1:-1] += dCdt * dt # Mettre a jour C

    #conditions des marges des concentrations
    C[0] = C[1] # condition bord gauche
    C[-1] = 0   # condition bord droit

    # plot
    if it % nout == 0:
        clear_output(wait=True) 
        ax.cla() 
        ax.plot(x, C)
        ax.set_ylim(0, 2000)
        ax.set_xlabel('Distance')
        ax.set_ylabel('Concentration')
        # CORRECTEUR: add a title in days 
        ax.set_title("Temps = " + str(int(time / d2s)) + " jours")
        # CORRECTEUR: add vertical lines to identify areas of initial high concentration and Venoge
        ax.plot([0, 0], [0, 2000], 'k')  # Vertical line at x=0
        ax.plot([x_init[0], x_init[0]], [0, 2000], 'r')  # Initial conditions
        ax.plot([x_init[1], x_init[1]], [0, 2000], 'r')  # Initial conditions
        ax.plot([venoge, venoge], [0, 2000], 'b')  # Veno line
        display(fig)
        plt.pause(0.1)
    
    # identification du premier dépassement 
    if C[int(venoge/dx)] > C_thr and flag_conc==True: # CORRECTEUR: use the threshold variable C_thr and convert venoge position to index
        #t_depassement = it*dt
        t_depassement = time/d2s  # CORRECTEUR: report time in days
        #flag = False # CORRECTEUR: you need to set flag_conc to False, not flag
        flag_conc = False

    # identification de la baisse de concentration dans la venoge
    if q_old[int(venoge/dx)] > q[int(venoge/dx)] and flag_flux==True:  # CORRECTEUR: use the stored venoge position variable and convert to index
        #t_baisse_flux = it*dt
        t_baisse_flux= time/d2s  # CORRECTEUR: report time in days
        flag_flux = False

#print("Temps du premier dépassement de la concentration pour la Venoge :", round(t_depassement/(86400), 2), "jours")
#print("Temps à la première baisse de flux :", round(t_baisse_flux/(86400), 2), "jours")
print(f"Temps de première contamination (>0.1 ppm): {t_depassement:.0f} jours")
print(f"Temps de début de diminution du flux: {t_baisse_flux:.0f} jours")
````
