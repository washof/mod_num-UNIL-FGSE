# Contamination d'un lac par des nutriments

Le but de cet exercice est de modéliser la contamination d'un lac par des nutriments (phosphore) provenant d'une station d'épuration défaillante. Le lac a une forme allongée et peut être modélisé en une dimension (1D) le long de son axe principal (Figure ci-dessous). Les nutriments se diffusent dans le lac et se dégradent naturellement par des processus biologiques.

**Vous allez construire un modèle numérique 1D** qui simule l'évolution de la concentration en phosphore $C(x,t)$ le long du lac (axe $x$) au cours du temps. Le but sera de prédire quand la concentration à une station de mesure atteindra un seuil critique.

Le modèle numérique sera basé sur une équation de diffusion-réaction pour le flux de concentration $q$ en fonction de la diffusivité $D$, de la concentration $C$ et de la distance $x$:

\begin{equation} 
  q = -D \frac{\partial C}{\partial x}. \qquad (1)
\end{equation}

L'équation ci-dessus peut alors être employée dans l'équation de conservation incluant un terme de dégradation biologique:

\begin{equation}
  \frac{\partial C}{\partial t} = -\frac{\partial q}{\partial x} - \gamma C. \qquad (2)
\end{equation}

où $\gamma$ est le taux de dégradation biologique.

Les deux équations doivent être discrétisées et implémentées dans un modèle numérique qui décrit l'évolution de la concentration en phosphore $C$ au cours du temps $t$ en se basant sur les paramètres donnés dans la table ci-dessous.

Vous allez construire un modèle numérique qui simule l'évolution de la concentration dans le lac de $x = 0$ m à $x = 5000$ m pendant 30 jours, en partant de $t = 0$. Le but sera de prédire quand la concentration à la station de mesure (à $x = 3500$ m) dépasse un certain seuil.

## Q2: Contamination intermittente

Pour la deuxième question, adaptez le code en introduisant un paramètre Q qui vaut 1 ou 2, afin que le code fonctionne pour les deux questions en choisissant `Q=1` ou `Q=2`.

Pour cette question, la pollution à 200 ppm ne se produit que entre 6h et 12h tous les jours. Pour savoir quelle heure il est, vous devez utiliser le temps $t$ (en jours) et déterminer l'heure de la journée. Par exemple, si $t = 2.25$ jours, cela correspond à 6h du troisième jour (0.25 jour = 6h).

**Aide** : Pour déterminer l'heure de la journée, vous pouvez :
1. Calculer la partie fractionnaire du temps : `heure_fraction = time - int(time)` (valeur entre 0 et 1)
2. Convertir en heure de la journée : `heure = heure_fraction * 24` (valeur entre 0 et 24)
3. Vérifier si l'`heure` pour savoir si la pollution se produit

````python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output, display

"""Version du code à effectuer"""
Q = 2

"""Paramètres physiques"""
D = 80000               # diffusivité de l'eau (en m2/jour)
x0 = 0                  # position entrée eau propre (en m)
xs = 1200               # position STEP (en m)
xm = 3500               # position station mesure (en m)
Lx = 5000               # longueur lac (en m)
Ci = 5                  # concentration initiale en phosphore (en ppm)
Cs = 200                # concentration à la STEP en phosphore (en ppm)
C0 = 5                  # concentration de l'eau propre en phosphore (en ppm)
Cm = 25                 # concentration seuil (en ppm)
gamma = 0.0006          # taux dégradation biologique (en 1/h)
ttot = 30               # temps total de la modélisation (en jours)
tstart = 6              # temps du début de la pollution intermittante (en heures)
tend = 12               # temps de la fin de la pollution intermittante (en heures)
h2d = 24                # facteur de correction heures en jour pour gamma (en h/j)

"""Paramètres numériques"""
nx = 501                      # nombre de cellules
dx = Lx / (nx - 1)            # taille d'une cellule
x = np.linspace(0, Lx, nx)    # coordonees des point du maillage
dt = dx**2 / (2.1 * D)        # condition sur le pas de temps
nt = int(ttot/dt)             # nombre de pas de temps
nout = 100                    # afficher la figure chaque 100e step

"""Initialisation"""
C = np.ones(nx) * Ci          # concentration initiale homogène
time = 0

# create the figure outside the loop 
fig, ax = plt.subplots()

"""Boucle temporelle"""
for it in range(nt):
    time += dt                                      # track time
    heure_fraction = time - int(time)               # calcul de l'heure de la journée
    heure = heure_fraction * h2d

    """Version 1 du code: flux constant en phosphore venant de la STEP"""
    if Q == 1:                          
        C[int(xs/dx)] = Cs                          # ajout de pollution constant de la STEP

    """Version 2 du code: flux intermittant"""
    if Q == 2 and heure >= tstart and heure <= tend:    # condition de pollution entre tstart et tend
            C[int(xs/dx)] = Cs                      # ajout de pollution de la STEP
    
    """Mise à jour des concentrations"""
    q = - D * ( C[1:] - C[:-1] ) / dx               # Mettre a jour le flux q en fonction de C 
    dCdt = - ( q[1:] - q[:-1] ) / dx                # Mettre a jour le flux q en fonction de C 
    C[1:-1] += dCdt * dt                            # Mettre a jour C
    
    dCdt_r = - gamma * h2d * C                      # Mise à jour de la réaction (avec gamma exprimé en 1/j au lieu de 1/h)
    C += dt*dCdt_r

    """Conditions de bord"""
    C[0] = C0           # arrivée d'eau propre en x=0
    C[-1] = C[-2]       # flux nul à droite

    """Condition de sortie"""
    if C[int(xm/dx)] >= Cm:
        print(f'La concentration à la position de la sonde ({xm} m) a dépassé {Cm} ppm après {round(time, 2)} jour(s).')
        break

    # plot
    if it % nout == 0:
        clear_output(wait=True) 
        ax.cla() 
        ax.set_title('Temps : ' + str(int(time)) + ' jours ' + str(int(heure)) + ' heures')
        ax.set_ylim([0, Cs])
        ax.set_xlim([0, Lx])
        ax.axvline(xs, label='Position STEP', linestyle='--', color='orange')
        ax.plot(xm, Cm, 'bo', label='Position sonde')
        ax.plot(x, C, 'r', label='Concentration')
        ax.set_xlabel('Position (m)')
        ax.set_ylabel('Concentration en phosphore (ppm)')
        ax.legend(loc='upper right')
        ax.grid(True, alpha=0.3)
        display(fig)
        plt.pause(0.1) 
````
