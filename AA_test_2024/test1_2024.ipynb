# ́Température dans le sol à Seattle

Le but de cet exercice est de modéliser l'évolution de la température dans le sol à un site particulier de Seattle, une ville située au Nord-Ouest des Etats-Unis, jusqu'à une profondeur de 30 m, comme illustré sur la figure ci-dessous. La température \(T(t,z)\) est ici une fonction du temps \(t\) (en années) et de la profondeur \(z\) (en mètres). La température initiale est de 10 °C sur toute la profondeur :

\begin{equation*} 
T(0,z) = 10^{\circ} C. \qquad (1)
\end{equation*}

A la surface ($z=30$ m), la température est donnée par la fonction suivante (condition de Dirichlet):

\begin{equation*} 
T (t,30) = - 2 - 10 \cos(2 \pi t), \qquad (2)
\end{equation*}

de sorte que la température oscille entre -12°C et 8°C au rythme des saisons. A la profondeur de 30 m ($z=0$), nous supposons qu'aucun flux de chaleur n'est possible à aucun moment (condition de Neumann): 

\begin{equation*} 
\frac{\partial T}{\partial z}(t,0) = 0. \qquad (3)
\end{equation*}

En plus des conditions de bord ci-dessus, on suppose la présence d'une conduite d'eau à une profondeur de 8 m sous la surface du sol ($z=22$ m), qui est active (alimentée en eau à 15°C) entre les temps $t_1=2$ ans et $t_2=7$ ans: 

\begin{equation*}
T (t,22) = 15^{\circ} C, \quad {\rm si} \quad t_1 \le t \le t_2. \qquad (4)
\end{equation*}

Le modèle numérique sera basé sur la loi de Fourier pour le flux de température \(q\) en fonction de la diffusivité \(D\), de la température \(T\) et de la distance verticale \(z\) :

\begin{equation*} 
  q = -D \frac{\partial T}{dz}, \qquad (5)
\end{equation*}

ainsi que l'équation traduisant la conservation de l'énergie:

\begin{equation*}
  \frac{\partial T}{\partial t} = -\frac{\partial q}{\partial z}. \qquad (6)
\end{equation*}

Les deux équations doivent être discrétisées et implémentées dans un modèle numérique qui décrit l'évolution temporelle de la température du sol \(T\), en se basant sur les paramètres donnés dans la table ci-dessous.

## Q2:  Deux couches (sédiments et rocheuse)

Pour la deuxième question, adaptez le code en introduisant un paramètre Q qui vaut 1 ou 2, afin que le code fonctionne pour pour les deux questions en choissant `Q=1`ouu `Q=2` (Si vous rencontrez un problème avec cela, vous avez toujours la possibilité de créer un nouveau bloc, copier-coller votre code de la premiere partie et l'adapter pour la question 2).

Pour cette question, au lieu d'une diffusivité $D$ homogène/constante dans l'espace, celle-ci a deux valeurs de $D$ différentes: $ D_{sed} = 15 $ m²/an au-dessus $z_{lim}=20$ m pour la couche de sédiments en haut et $ D_{roch} = 30 $ m²/an  au-dessous $z_{lim}=20$ m pour la couche rocheuse en bas. 

**Note:** prenez bien en compte pour modifier le code que $D$ est maintenant un vecteur après discrétisation, ainsi que l'adaptation du pas de temps : $ dt = dz^2 / (2.1 \times max( D_{sed}, D_{roc})) $. Attention à ce que le vecteur $D$ ait la bonne dimension.


**Question 2** : Comme dans la question 1, combien de temps s'écoule-t-il jusqu'à ce que la température à la ligne de contrôle (z = 10 m) enregistre une température inférieure à une valeur seuil de 8 °C ?

````python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, clear_output

# Choix de la question (1 ou 2)
Q = 1

# ========================================
# PARAMÈTRES PHYSIQUES
# ========================================
zsurf    = 30       # altitude de la surface, m
zcond    = 22       # profondeur de la conduite d'eau, m
zlim     = 20       # profondeur de transition entre couches (Q2), m
zcont    = 10       # profondeur de la ligne de contrôle, m
zbase    = 0        # altitude du bord inférieur, m
Tinit    = 10       # température initiale du sol, °C
Tcond    = 15       # température de la conduite d'eau, °C
Dsed     = 15       # diffusivité des sédiments, m²/an
t1       = 2        # début d'activation de la conduite, an
t2       = 7        # fin d'activation de la conduite, an
ttotal   = 100      # temps total de simulation, an
Tseuil   = 8        # seuil de température à la ligne de contrôle, °C
time     = 0        # temps initial, an

# ========================================
# PARAMÈTRES NUMÉRIQUES
# ========================================
nz     = 101                            # nombre de points de grille
dz     = zsurf / (nz - 1)               # pas d'espace, m
z      = np.linspace(zbase, zsurf, nz)  # coordonnées verticales, m
nout   = 100                            # intervalle de sortie graphique
 
# Configuration selon la question
if Q == 1:
    # Question 1 : une seule couche homogène
    D  = Dsed                             # diffusivité constante, m²/an
    dt = dz**2 / (2.1 * Dsed)             # pas de temps (stabilité), an
elif Q == 2:
    # Question 2 : deux couches (sédiments + roche)
    D     = Dsed * np.ones(nz - 1)        # vecteur de diffusivité (nz-1 faces)
    Droch = 30                            # diffusivité de la roche, m²/an
    dt    = dz**2 / (2.1 * max(Dsed, Droch))  # pas de temps (stabilité), an
    # Attribution de la diffusivité rocheuse en-dessous de zlim
    D[:int(zlim / zsurf * nz)] = Droch

nt = int(ttotal / dt)   # nombre total d'itérations temporelles

# ========================================
# INITIALISATION
# ========================================
T = np.ones(nz) * Tinit  # température initiale uniforme, °C

# Configuration de la figure avec style amélioré
fig, ax = plt.subplots(figsize=(8, 6)) 

# ========================================
# BOUCLE TEMPORELLE
# ========================================
for it in range(1, nt):

    # Avancement du temps
    time += dt
    
    # Calcul du flux de chaleur q = -D * dT/dz (loi de Fourier, éq. 5)
    # Différence finie : dT/dz ≈ (T[i+1] - T[i]) / dz
    q = -D * (T[1:] - T[:-1]) / dz
    
    # Calcul de la variation de température dT/dt = -dq/dz (conservation, éq. 6)
    # Différence finie : dq/dz ≈ (q[i+1] - q[i]) / dz
    dTdt = -(q[1:] - q[:-1]) / dz
    
    # Mise à jour de la température aux points intérieurs 
    T[1:-1] += dt * dTdt

    # ========================================
    # CONDITIONS AUX LIMITES
    # ========================================
    # Condition de Neumann au fond : flux nul, dT/dz = 0 (éq. 3)
    T[0] = T[1]
    
    # Condition de Dirichlet en surface : température oscillante (éq. 2)
    T[-1] = -2 - 10 * np.cos(time * 2 * np.pi)

    # ========================================
    # CONTRAINTE DE LA CONDUITE D'EAU (éq. 4)
    # ========================================
    if (t1 < time) & (time < t2):
        # Imposer la température de la conduite entre t1 et t2
        T[int(zcond / dz)] = Tcond

    # ========================================
    # VÉRIFICATION DU SEUIL DE TEMPÉRATURE
    # ========================================
    if T[int(zcont / dz)] < Tseuil:
        print(f'La température est passée sous {Tseuil}°C à {zcont} m après {time:.2f} ans')
        break

    # ========================================
    # AFFICHAGE GRAPHIQUE
    # ========================================
    if it % nout == 0:
        clear_output(wait=True)
        ax.cla()
        
        # Tracé du profil de température
        ax.plot(T, z, linewidth=2.5, color='orange', label='Température')
        
        # Marqueurs importants
        ax.axhline(zcond, color='blue', linestyle='--', linewidth=1.5, 
                   label=f'Conduite d\'eau ({zcond} m)')
        ax.axhline(zcont, color='green', linestyle='--', linewidth=1.5, 
                   label=f'Ligne de contrôle ({zcont} m)')
        if Q == 2:
            ax.axhline(zlim, color='orange', linestyle=':', linewidth=1.5, 
                       label=f'Transition couches ({zlim} m)')
        
        # Annotations et mise en forme
        ax.set_title(f'Profil de température à t = {time:.2f} ans (Q{Q})', 
                     fontsize=14, fontweight='bold')
        ax.set_xlabel('Température (°C)', fontsize=12)
        ax.set_ylabel('Profondeur (m)', fontsize=12)
        ax.set_xlim([-15, 20])
        ax.set_ylim([zbase, zsurf])
        ax.grid(True, alpha=0.3)
        ax.legend(loc='lower left', fontsize=10)
        
        # Affichage
        display(fig)
        plt.pause(0.1)

plt.close()
````
